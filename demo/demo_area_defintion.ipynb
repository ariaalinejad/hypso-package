```python
from pyproj import Transformer
from pyproj import Proj
from pyresample import geometry

area_id = 'lacrau'
proj_id = 'roi'
description = 'roi'

# take center of interest 43°33'32.0"N 4°51'51.0"E 
# convert to decimal degrees: 43 + (33/60) + (32/3600) ≈ 43.5589°N , 4 + (51/60) + (51/3600) ≈ 4.8642°E

lon_0 = 4.8642
lat_ts = 43.5589
buffer_lon = 2
buffer_lat = 3

# Updated bbox centered around 43.5589°N, 4.8642°E with 0.2° buffer
bbox = (lon_0 - buffer_lon, lat_ts - buffer_lat, lon_0 + buffer_lon, lat_ts + buffer_lat) # (min_lon, min_lat, max_lon, max_lat)

# lon_0 = (bbox[2] - bbox[0])/2 + bbox[0]
# lat_ts = (bbox[3] - bbox[1])/2 + bbox[1]

# Stereographic projection centered at the North Pole (lat_0=90.0 means polar stereographic projection, different projections might be better for different regions)
p = Proj(proj="stere", ellps="bessel", lat_0=90.0, lon_0=lon_0, lat_ts=lat_ts)

projection = {
    "proj": "stere",
    "ellps": "bessel",
    "lat_0": 90.0,
    "lon_0": lon_0,
    "lat_ts": lat_ts,
    "units": "m"
}

# Calculate area extent in projected coordinates
lower_left_x, lower_left_y = p(bbox[0], bbox[1])
upper_right_x, upper_right_y = p(bbox[2], bbox[3])
area_extent = (lower_left_x, lower_left_y, upper_right_x, upper_right_y)

res = 300  # meters

# Compute width and height in pixels
width = int((area_extent[2] - area_extent[0]) / res)
height = int((area_extent[3] - area_extent[1]) / res)

# Define the area
area_def = geometry.AreaDefinition(
    area_id, proj_id, description, projection, width, height, area_extent
)
```

## Save area definition to file

```python
area_def.dump(filename='./'+area_id+'_demo.yaml')
```
